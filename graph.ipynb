```python
from matplotlib import pyplot as plt
from matplotlib.patches import FancyBboxPatch

def add_fancy_box(ax, x, y, width, height, text, fontsize=10, color='white', edgecolor='black'):
    """Add a box with text inside."""
    box = FancyBboxPatch((x, y), width, height, boxstyle="round,pad=0.3", edgecolor=edgecolor, linewidth=1.5, facecolor=color)
    ax.add_patch(box)
    ax.text(x + width/2, y + height/2, text, fontsize=fontsize, ha="center", va="center", color='black', zorder=5)

def draw_gru_model_architecture():
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.axis('off')
    ax.set_xlim(0, 12)
    ax.set_ylim(0, 10)
    
    # GRU Layers
    add_fancy_box(ax, 1, 8, 2, 1, "Input\n(559 x ?)", color="#d5e8d4")
    add_fancy_box(ax, 4, 8, 2, 1, "Bidirectional GRU\n(559 x 256)", color="#dae8fc")
    add_fancy_box(ax, 7, 8, 2, 1, "BatchNorm\n(559 x 256)", color="#fff2cc")
    add_fancy_box(ax, 10, 8, 2, 1, "Dropout\n(559 x 256)", color="#f8cecc")

    # Connections between layers (First Layer)
    ax.arrow(3, 8.5, 0.8, 0, head_width=0.2, head_length=0.2, fc="black", ec="black")
    ax.arrow(6, 8.5, 0.8, 0, head_width=0.2, head_length=0.2, fc="black", ec="black")
    ax.arrow(9, 8.5, 0.8, 0, head_width=0.2, head_length=0.2, fc="black", ec="black")

    # Second GRU Layer
    add_fancy_box(ax, 4, 6, 2, 1, "Bidirectional GRU\n(559 x 128)", color="#dae8fc")
    add_fancy_box(ax, 7, 6, 2, 1, "BatchNorm\n(559 x 128)", color="#fff2cc")
    add_fancy_box(ax, 10, 6, 2, 1, "Dropout\n(559 x 128)", color="#f8cecc")
    
    # Connections for second GRU
    ax.arrow(5, 7.5, 0, -0.5, head_width=0.2, head_length=0.2, fc="black", ec="black")
    ax.arrow(6, 6.5, 0.8, 0, head_width=0.2, head_length=0.2, fc="black", ec="black")
    ax.arrow(9, 6.5, 0.8, 0, head_width=0.2, head_length=0.2, fc="black", ec="black")

    # Third GRU Layer
    add_fancy_box(ax, 4, 4, 2, 1, "Bidirectional GRU\n(64)", color="#dae8fc")
    add_fancy_box(ax, 7, 4, 2, 1, "BatchNorm\n(64)", color="#fff2cc")
    add_fancy_box(ax, 10, 4, 2, 1, "Dropout\n(64)", color="#f8cecc")

    # Connections for third GRU
    ax.arrow(5, 5.5, 0, -0.5, head_width=0.2, head_length=0.2, fc="black", ec="black")
    ax.arrow(6, 4.5, 0.8, 0, head_width=0.2, head_length=0.2, fc="black", ec="black")
    ax.arrow(9, 4.5, 0.8, 0, head_width=0.2, head_length=0.2, fc="black", ec="black")

    # Dense Layers
    add_fancy_box(ax, 6, 2, 2, 1, "Dense\n(16)", color="#f5f5f5")
    ax.arrow(7, 3.5, 0, -0.5, head_width=0.2, head_length=0.2, fc="black", ec="black")
    add_fancy_box(ax, 6, 0.5, 2, 1, "Dense\n(2)", color="#f5f5f5")
    ax.arrow(7, 1.5, 0, -0.5, head_width=0.2, head_length=0.2, fc="black", ec="black")

    plt.title("GRU Model Architecture", fontsize=16)
    plt.show()

draw_gru_model_architecture()
```
